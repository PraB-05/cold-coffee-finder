```python
from geopy.distance import geodesic
```

```python
spot_1 = (25.173230,55.280982)
round(geodesic(user_locations,spot_1).km,2)
```

```python
def calculate_distance(user_locations,spot_location):
    return round(geodesic(user_locations,spot_location).km,2)
```

```python
def get_distance(row):
    lat = row['lat']
    lng = row['lng']
    spot = (lat , lng)
    return calculate_distance(user_locations, spot)
```

```python
df['distance'] = df.apply(get_distance,axis = 1)
```

```python
from datetime import datetime
current_time = datetime.now().strftime("%H:%M")
current_time
```

```python
def is_open_spot(row):
    open_time = row['opening_time']
    close_time = row['closing_time']
    if open_time <= current_time<= close_time:
        return "Open"
    else:
        return 'Closed'
```

```python
df['is_open_spot'] = df.apply(is_open_spot , axis = 1)
```
